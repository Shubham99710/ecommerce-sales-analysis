***Cleaning & Transforming Raw Data (Python & Pandas)***

```python
import pandas as pd

# Loading raw data with specified encoding
df = pd.read_csv("Online_Retail.csv", encoding='latin1')

# Inspecting the structure
print(df.info())
print(df.head())
```

```python
# Checking missing value counts per column
print(df.isnull().sum())

# Drop rows where CustomerId is missing
df = df.dropna(subset = ["CustomerID"])

# Cast CustomerId to integer
df["CustomerID"] = df["CustomerID"].astype(int)
```

```python
# Removing tansactions with zero or negative quantities/prices
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
```

```python
# Converting strings to actual Datetime objects, specifying the format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], format="%m/%d/%y %H:%M")

# Creating a Total Sales column
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]

#Extracting Month and Year for grouping later
df["InvoiceMonth"] = df["InvoiceDate"].dt.to_period("M")
```

```python
df.to_csv("cleaned_retail.csv", index = False)
print("Data cleaning complete. File saved as cleaned_retail.csv")
```
